**Task 1**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Dataset .csv")
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
table_booking = df["Has Table booking"].value_counts()

table_booking_percentage = (table_booking / len(df)) * 100

table_booking_percentage
```

```python
online_delivery = df["Has Online delivery"].value_counts()

online_delivery_percentage = (online_delivery / len(df)) * 100

online_delivery_percentage
```

```python
table_booking_rating = df.groupby("Has Table booking")["Aggregate rating"].mean()

table_booking_rating
```

```python
table_booking_rating.plot(kind="bar")

plt.show()
```

```python
delivery_price = pd.crosstab(df["Price range"],df["Has Online delivery"])

delivery_price
```

```python
delivery_price.plot(kind="bar")

plt.show()
```
